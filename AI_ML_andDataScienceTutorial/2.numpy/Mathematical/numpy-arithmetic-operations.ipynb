```python
# addition of arrays
import numpy as np
a= np.array([2,3,4,5])
b = np.array([2,3,4,5])
res = np.add(a,b)
print(list(res))
print(res)
```

```python

#subtract
res1 = np.subtract(a,b)
print(res1)
```

```python
#multiply
res2 = np.multiply(a,b)
print(res2)
```

```python
#divide
res3 = np.divide(a,b)
print(res3)
print(list(res3))
```
