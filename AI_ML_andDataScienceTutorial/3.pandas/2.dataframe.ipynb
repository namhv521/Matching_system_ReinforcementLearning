```python
# 2.7 merging joining concatenating
# pd.concat([df1,df2]) nối các df theo hang axis = 0 hoac theo cot axis = 1
# pd.merger(df1,df2, on='key' ,how = '.') noi theo khoa giong join trong sql
# cac loai join, ineer, left ,right, outer, cross

khach_hang = pd.DataFrame({
    'ma_kh': [1, 2, 3],
    'ten_kh': ['An', 'Binh', 'Chi']
})
don_hang = pd.DataFrame({
    'ma_kh': [1, 2, 2, 4],
    'san_pham': ['Bút', 'Vở', 'Thước', 'Gôm']
})
# inner join chir giữ khớp cả 2 bên
print(pd.merge(khach_hang, don_hang, on='ma_kh',how='inner'))

# left hoin giu het ben trai
print(pd.merge(khach_hang, don_hang, on='ma_kh', how='left'))

#right join giu het phan ben phai
print(pd.merge(khach_hang, don_hang, on='ma_kh', how='right'))

#outer join, giu het ca 2 ben
print(pd.merge(khach_hang, don_hang, on='ma_kh', how='outer'))

# su dung concat de noi df
khach_hang_tmp = pd.DataFrame({
    'ma_kh':[4], 'ten_kh':['Dung']
})
print(pd.concat([khach_hang,khach_hang_tmp],ignore_index=True))
```
