```python
# local variables
def greet():
    msg = " hello from inside the function!"
    print(msg)

greet()
```

```python
# global variables
msg= "python is awesome!"

def display():
    print("inside function: ",msg)

display()
print("outside function: ", msg)
```

```python
# use of local and global variables
def fun():
    s= " me too"
    print(s)

s = " i love geeksforgeeks"
fun()
print(s)
```

```python
# modifying global variables inside a function
s = "i love gfg"

def fun():
    global s
    s += 'gfg'
    print(s)
    s = "look for geg python section"
    print(s)

fun()
print(s)
```

```python
# global vs local with same name
a= 1
def f():
    print("f(a): ",a)
    b = a
    print("f(b): ",b)

def g():
    a = 2
    print("g(): ",a)

def h():
    global a
    a = 3
    print("h(): ",a)

print("global: ",a)
f()
print("global: ",a)
g()
print("global: ",a)
h()
print("global: ",a)
```
