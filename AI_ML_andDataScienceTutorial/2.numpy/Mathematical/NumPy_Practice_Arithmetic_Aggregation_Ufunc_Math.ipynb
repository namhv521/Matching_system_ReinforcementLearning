# NumPy Practice — Arithmetic, Aggregation, UFuncs & Mathematical Functions

````python
import numpy as np
````

## Phần 1 — Basic Arithmetic Operations


### Bài 1 — Phép toán giữa hai mảng
Cho:
```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])
```
Hãy tính:
1. `a + b`
2. `a - b`
3. `a * b`
4. `a / b`
5. `a ** b`
6. `a % b`


### Bài 2 — Phép toán với scalar
Cho `x = np.array([2, 4, 6, 8, 10])`.
Tính `x + 5`, `x - 3`, `x * 4`, `x / 2`, `x ** 2`.


### Bài 3 — Arithmetic trên ma trận
Cho:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[6, 5, 4],
              [3, 2, 1]])
```
Thực hiện `A+B`, `A-B`, `A*B`, `A/B` và `A**2`.
Sau đó giải thích ngắn gọn sự khác nhau giữa `A * B` và phép nhân ma trận.


### Bài 4 — Broadcasting
Cho ma trận:
```python
X = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])
```
1. Cộng `[1, 2, 3]` vào từng hàng.
2. Nhân mỗi cột lần lượt với `2, 3, 4`.
3. Trừ `[10, 20, 30]` khỏi từng hàng.


## Phần 2 — Aggregation Functions

- Tổng điểm

### Bài 7 — Phân tích điểm sinh viên
Cho:
```python
scores = np.array([
    [8, 7, 9, 6],
    [5, 8, 7, 9],
    [9, 9, 10, 8],
    [6, 7, 5, 8],
    [10, 8, 9, 9]

```
Mỗi hàng là một sinh viên, mỗi cột là một môn.
1. Tìm điểm trung bình của từng sinh viên.
2. Tìm điểm trung bình của từng môn.
3. Tìm sinh viên có điểm cao nhất và thấp nhất dựa trên điểm trung bình.
4. Tìm điểm cao nhất của từng môn.
5. Tìm điểm thấp nhất của từng môn.


### Bài 8 — `uint8` và overflow
Cho:
```python
arr = np.array([[14, 17, 12, 33, 44],
                [15,  6, 27,  8, 19],
                [23,  2, 54,  1,  4]])
```
Tính:
```python

```

## Phần 3 — Universal Functions (ufunc)


### Bài 9 — Ufunc cơ bản
Cho `x = np.array([-5, -2, 0, 3, 7])`.
Dùng các ufunc thích hợp để tính:
- Giá trị tuyệt đối
- Bình phương
- Căn bậc hai của các giá trị không âm
- Làm tròn xuống
- Làm tròn lên

Gợi ý: tìm hiểu `np.abs`, `np.square`, `np.sqrt`, `np.floor`, `np.ceil`.

### Bài 10 — Ufunc với hai mảng
Cho:
```python

b = np.array([3, 4, 2, 5])
```
Dùng ufunc để tính từng phần tử của:
- `a + b`
- `a * b`
- `a ** b`
- `max(a, b)` theo từng vị trí
- `min(a, b)` theo từng vị trí


### Bài 11 — Kiểm tra ufunc

```python
np.add

```
Sử dụng thuộc tính/cách kiểm tra thích hợp của NumPy và ghi nhận kết quả.

### Bài 12 — `out` trong ufunc


Sau đó kiểm tra `result`.

### Bài 18 — Chuẩn hóa Min-Max
Cho:
```python
x = np.array([10, 20, 35, 50, 80, 100])
```
Áp dụng công thức Min-Max Scaling:

`x_scaled = (x - min(x)) / (max(x) - min(x))`

Yêu cầu:
1. Không dùng vòng lặp.
2. Chỉ sử dụng NumPy.
3. Kiểm tra giá trị nhỏ nhất và lớn nhất của mảng sau khi chuẩn hóa.

# Gợi ý kiến thức cần nhớ

Trước khi làm bài, nên nắm chắc:
- Arithmetic: `+`, `-`, `*`, `/`, `//`, `%`, `**`

- `axis=0`, `axis=1`
- Ufunc: `np.add`, `np.subtract`, `np.multiply`, `np.divide`, `np.sqrt`, `np.abs`, `np.maximum`, `np.minimum`, ...
