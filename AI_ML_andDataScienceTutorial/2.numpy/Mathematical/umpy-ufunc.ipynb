```python
# các giá trị cần chú ý trong xác suất thống kê như mean, median, max, min, std(độ lệch chuẩn), var( phương sai)
import numpy as np
a = np.array([1,2,34,5,6,7])
print("min: " , np.amin(a))
print("max: ", np.amax(a))
print("std: ", np.std(a))
print("var: ",np.var(a))
print("mean: ",np.mean(a))
print("median: ",np.median(a))
```
