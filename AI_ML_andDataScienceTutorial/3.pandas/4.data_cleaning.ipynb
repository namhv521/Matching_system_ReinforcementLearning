```python
#1.2 loai bo du lieu thieu - dropna()
# df.dropna(axis=0, how= ' any,thresh=None,subset=None, inplace=False)

df=pd.read_csv('data/employees.csv')
#xoa toan bo cac hang co it nhat 1 gia tri thieu
df_drop_any = df.dropna(how='any')
print("so dong con lai: ", len(df_drop_any),"/",len(df),'\n')

#chi oa hang neu thieu o cot quan trong 
df_drop_subset = df.dropna(subset=['First Name'])
# print(df_drop_subset,'\n')
print("so dong con lai", len(df_drop_subset),'\n')

#giu lai cac hang co it nhat 6 gia tri khong Nan( tren tong 8 cot)
df_thresh = df.dropna(thresh = 6)
print("so: ", len(df_thresh),'\n')
```

```python
# 1.3 dien gia tri con thieu - fillna()
# df.fillna(value = None, method =None, axis = None, limit = None)


#dien gia tri con thieu cho cot phan loai(categorical) bang 1 gia tri co dinh
df_fill = df.copy()
print(df_fill.head(20),'\n')
df_fill['Team'] = df_fill['Team'].fillna('Unknown')
df_fill['Gender'] = df_fill['Gender'].fillna('Not specified')
df_fill[['Team','Gender']].isna().sum()
```

```python
#dien gi tri con thieu cho cot so bang mean or medina
df_fill2 = df.copy()
mean_salary = df_fill2['Salary'].mean()
df_fill2['Salary'] = df_fill2['Salary'].fillna(mean_salary)
# df_fill2.head(20)
print("luong trung binh dung de dien: ", round(mean_salary,2),'\n')
```

```python
#fillna voi method='ffill' lay gia tri lien tuoc va 'bfill' lay gia tri lien sau
import numpy as np
demo = pd.Series([1,np.nan,np.nan,np.nan,4,np.nan,6])
print("goc: ", demo.tolist,'\n')
print("ffill: ",demo.ffill().tolist,'\n')
print("bfill: ",demo.bfill().tolist(),'\n')
```

```python
# noi suy gia tri bi thieu
# interpolate()
demo2 = pd.Series([10,np.nan,np.nan,40,50])
print("goc:     ",demo2.tolist())
print("noi suy gia tri yeu: ",demo2.interpolate(method = 'linear').tolist())
```
