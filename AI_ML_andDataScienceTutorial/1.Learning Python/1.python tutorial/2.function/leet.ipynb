```python
#roman-to-integer/
class Solution:
    def romanToInt(self, s: str) -> int:
        roman = {'I':1 , 'V':5 ,'X':10 , 'L':50, 'C':100,
        'D':500, 'M':1000 }

        res = 0
        for i in range(len(s)):
            curr= roman[s[i]]
            next_val = roman[s[i+1]] if i+1<len(s) else 0

            if curr < next_val:
                res -= curr
            else:
                res += curr

        return res
```

```python
# https://leetcode.com/problems/longest-common-prefix/
"""
ý tưởng là sort lại mảng ban đầu khi đó chỉ cần so sánh cái ầu tiên và cái cuối cùng là được
"""
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        ans = ""
        strs = sorted(strs)
        first = strs[0]
        last = strs[-1]
        for i in range(min(len(first), len(last))):
            if first[i] != last[i]:
                return ans
            ans += first[i]

        return ans
```

```python
# https://leetcode.com/problems/valid-parentheses/
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        start = ['[', '(', '{']
        end = [']', ')', '}']

        if len(s) % 2 != 0:
            return False

        for i in s:
            if i in start:
                stack.append(i)  # gặp dấu mở -> đẩy vào stack
            else:
                # gặp dấu đóng -> phải có dấu mở tương ứng ở đỉnh stack
                if not stack:
                    return False
                pop_value = stack.pop()
                get_index_start = start.index(pop_value)
                if i != end[get_index_start]:
                    return False

        return not stack  # stack phải rỗng thì mới hợp l
```
