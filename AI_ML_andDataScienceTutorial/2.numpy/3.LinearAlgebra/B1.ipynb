```python
# b2
def create_matrix(rows, cols, fill_value):
    a = np.full((rows,cols),fill_value)
    return a
full_7 = create_matrix(3,4,7)
print(full_7)
full_7_T = full_7.T
print(full_7_T)

resh = full_7.reshape(4,3)
print(resh)
```
