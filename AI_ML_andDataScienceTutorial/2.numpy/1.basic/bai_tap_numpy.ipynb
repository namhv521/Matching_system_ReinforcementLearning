## Đáp án tham khảo (gợi ý ngắn gọn)

- **1.3**: `a.dtype` là `float32`, `b.dtype` là `int64` (hoặc `int32` tuỳ hệ điều hành). Kết quả `a+b` có dtype float vì NumPy tự "upcast" kiểu dữ liệu nhỏ hơn lên kiểu lớn hơn để không mất thông tin (type promotion).
- **3.4**: `a` sẽ bị thay đổi thành `[0,1,99,3,4,5,6,7,8,9]` vì slicing cơ bản (`a[2:5]`) trả về **view**, không phải bản sao — `b` và `a` dùng chung vùng nhớ.
- **4.3**: `np.resize` khi kích thước mới lớn hơn sẽ **lặp lại (tile) dữ liệu gốc** để lấp đầy, trong khi `reshape` sẽ báo lỗi nếu tổng số phần tử không khớp.
- **4.4**: mảng gốc có 6 phần tử `(2,3)`, còn `(4,2)` cần 8 phần tử → không khớp. Shape hợp lệ: `(3,2)` hoặc `(6,1)` hoặc `(1,6)`...
- **5.3**: `concatenate` nối các mảng theo trục có sẵn, **không tăng số chiều** của kết quả so với input; `stack` luôn **tạo thêm một trục mới**.
- **5.4**: `np.split` yêu cầu chia đều tuyệt đối, 9 phần tử không chia hết cho 4 nên sẽ báo lỗi; `array_split` cho phép chia không đều (một số phần sẽ có nhiều hơn 1 phần tử).
- **6.3**: `(a+b).shape=(3,4)`, `(a+c).shape=(3,4)`, `(b+c).shape=(3,4)` — dòng cuối: `b` shape `(4,)` được coi như `(1,4)`, `c` shape `(3,1)`, cả hai cùng "giãn" theo chiều thiếu để khớp thành `(3,4)`.
- **6.4**: `(3,4)` và `(3,)` không tương thích vì so từ phải sang trái: `4` vs `3` không khớp và không có chiều nào bằng 1. Sửa `b` thành `(4,)` để cộng theo hàng, hoặc `(3,1)` để cộng theo cột.
