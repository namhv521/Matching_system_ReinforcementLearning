```python
import numpy as np 
import pandas as pd
```

```python
#series la di lieu 11 chieu, co nhan index di kem nhu 1 cot trong bang tinh
# tao 1 series
# cos the tao series tu list sict,numpy array, hoac 1 gia ri scalar


# tao tu list -> index mac inh la 0,1,2
s1 = pd.Series([10,20,30,40])
print(s1)

# tao tu dict (bao gom key:value) -> key tro thanh index 
s2 = pd.Series({'a':1, 'b':2, 'c':3})
print(s2)

# tao voi index tuy chinh
s3 = pd.Series ([100,200,300],index = ['x','y','z'])
print(s3)
```

```python
#series luu diem 5 mon hoc voi index la ten mon
diem = pd.Series([7,6,4,8.5,8],index =['toan','ly','hoa','van','anh'])
print(diem)
print("diem trung binh:", diem.mean())
```

```python
# bai 1.1 
danso=pd.Series([10,20,15,2,5],index=['Ha noi','Thanh Hoa','Nam Dinh','Ha Nam','Phu tho'])
res = danso.idxmax()
print(res)

# danso.max() la in ra value max
# danso.idxmax() la in ra index co value max

```

```python
#1.2 truy cap phan tu cua series
s = pd.Series([10,20,30,40,50],index = ['a','b','c','d','e'])
print(s['b']) # theo nhan
print(s.loc['b']) # tuong duong.loc
print(s.iloc[1]) # theo vi tri( so 1, phan tu thu 2)
print(s[['a','c']]) #nhieu phan tu cung 1 luc
print(s[1:3]) #slicing theo vi tri [1:3)
```

```python
""" 1.3 cac phuong thuc lam viec voi index cua series

s.index.tolist() :  xem index hien tai
s.reindex([..]) sap xep, doi lai index, gia tij khong khop se thanh Nan
s.reset_index dua index cu thanh 1 co,t gan index moi
s.rename doi ten index
"""
s = pd.Series([11,2,3],index = ['a','b','c'])
print("index", s.index.tolist())  # xem danh sach cac index hien tai
s_reindexed = s.reindex(['a','b','c','d']) #-> 'd' chua co, nen value thanh Nan
print(s_reindexed)

print(s.reset_index())   #
```

```python
# dung reindex de them 2 tinh moi chua co du lieu vao series danso sau do fung fillna(0) de dien gia tri 0 cho cac tinh moi nay
# danso_new =danso.reindex(['Ha noi','Thanh Hoa','Nam Dinh','Ha Nam','Phu tho','Hung Yen','Thai Binh'])
# danso_new = danso_new.fillna(0)
danso_new =danso.reindex(['Ha noi','Thanh Hoa','Nam Dinh','Ha Nam','Phu tho','Hung Yen','Thai Binh','Hai phong'],fill_value=1)
print(danso_new)
```

```python
#1.4 nhi phan tren series
"""khi su dung cac phep toan giua 2 series thi chung se tu dong can chinh theo index
neu index khong khpp thif ket qua se la Nan"""
a = pd.Series([1,2,3],index=['x','y','z'])
b = pd.Series([10,20,30],index = ['z','y','w'])
print(a+b) # index khong khop het -> co Nan
print(a.add(b,fill_value=0)) # dung fill_value de tranh Nan
print(a*2)
```
