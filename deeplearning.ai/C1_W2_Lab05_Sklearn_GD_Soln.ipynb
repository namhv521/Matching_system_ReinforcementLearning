## Công cụ

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from lab_utils_multi import  load_house_data
from lab_utils_common import dlc
np.set_printoptions(precision=2)
plt.style.use('./deeplearning.mplstyle')
```

# Gradient Descent
Scikit-learn có một mô hình hồi quy gradient descent là [sklearn.linear_model.SGDRegressor](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.SGDRegressor.html#examples-using-sklearn-linear-model-sgdregressor). Giống như cách triển khai gradient descent trước đây của bạn, mô hình này hoạt động tốt nhất với đầu vào đã được chuẩn hóa. [sklearn.preprocessing.StandardScaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler) sẽ thực hiện chuẩn hóa z-score giống như trong lab trước. Ở đây nó được gọi là 'standard score' (điểm chuẩn hóa).

### Nạp tập dữ liệu

```python
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

### Chuẩn hóa dữ liệu huấn luyện

```python
scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
print(f"Peak to Peak range by column in Raw        X:{np.ptp(X_train,axis=0)}")   
print(f"Peak to Peak range by column in Normalized X:{np.ptp(X_norm,axis=0)}")
```

### Tạo và khớp mô hình hồi quy

```python
sgdr = SGDRegressor(max_iter=1000)
sgdr.fit(X_norm, y_train)
print(sgdr)
print(f"number of iterations completed: {sgdr.n_iter_}, number of weight updates: {sgdr.t_}")
```

### Xem các tham số
Lưu ý, các tham số này gắn liền với dữ liệu đầu vào *đã chuẩn hóa*. Các tham số khớp được rất gần với những tham số tìm thấy ở lab trước với cùng dữ liệu này.

```python
b_norm = sgdr.intercept_
w_norm = sgdr.coef_
print(f"model parameters:                   w: {w_norm}, b:{b_norm}")
print( "model parameters from previous lab: w: [110.56 -21.27 -32.71 -37.97], b: 363.16")
```

### Thực hiện dự đoán
Dự đoán giá trị mục tiêu của dữ liệu huấn luyện. Sử dụng cả hàm `predict` và tính toán bằng $w$ và $b$.

```python
# thực hiện dự đoán bằng sgdr.predict()
y_pred_sgd = sgdr.predict(X_norm)
# thực hiện dự đoán bằng w,b. 
y_pred = np.dot(X_norm, w_norm) + b_norm  
print(f"prediction using np.dot() and sgdr.predict match: {(y_pred == y_pred_sgd).all()}")

print(f"Prediction on training set:\n{y_pred[:4]}" )
print(f"Target values \n{y_train[:4]}")
```

### Vẽ đồ thị kết quả
Hãy vẽ đồ thị dự đoán so với giá trị mục tiêu.

```python
# vẽ dự đoán và giá trị mục tiêu theo các đặc trưng gốc    
fig,ax=plt.subplots(1,4,figsize=(12,3),sharey=True)
for i in range(len(ax)):
    ax[i].scatter(X_train[:,i],y_train, label = 'target')
    ax[i].set_xlabel(X_features[i])
    ax[i].scatter(X_train[:,i],y_pred,color=dlc["dlorange"], label = 'predict')
ax[0].set_ylabel("Price"); ax[0].legend();
fig.suptitle("target versus prediction using z-score normalized model")
plt.show()
```
