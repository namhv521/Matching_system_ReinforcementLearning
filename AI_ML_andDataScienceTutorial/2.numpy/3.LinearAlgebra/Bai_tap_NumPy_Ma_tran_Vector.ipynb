## CHỦ ĐỀ 1: MATRIX MULTIPLICATION & MANIPULATION

### A. Đọc code – Trả lời tự luận

**A1 (Nhận biết)**

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A * B
D = A @ B
print(C)
print(D)
```

```python
import numpy as np
A = np.arange(1, 10).reshape(3, 3)
B = A.T
C = np.matmul(A, B)
print(A)
print(B)
print(C)
print(C == C.T)
```

## CHỦ ĐỀ 3: DETERMINANT & INVERSE OF A MATRIX

### A. Đọc code – Trả lời tự luận

**A1 (Nhận biết)**

```python
import numpy as np
A = np.array([[4, 7], [2, 6]])
det_A = np.linalg.det(A)
print(det_A)
```

Hỏi: Hàm nào được dùng để tính định thức? Với ma trận 2x2 `[[a,b],[c,d]]`, định thức được tính bằng công thức nào, và hãy kiểm chứng bằng tay kết quả trên.

**A2 (Thông hiểu)**

```python
import numpy as np
A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)
print(A_inv)
print(np.round(A @ A_inv))
```

Hỏi: Dòng cuối in ra kết quả gì và tại sao? Nêu mối liên hệ giữa ma trận nghịch đảo và ma trận đơn vị (identity matrix).

**A3 (Vận dụng)**

```python
import numpy as np
B = np.array([[1, 2], [2, 4]])
try:
    B_inv = np.linalg.inv(B)
except np.linalg.LinAlgError as e:
    print("Loi:", e)
print(np.linalg.det(B))
```

Hỏi: Vì sao đoạn code trên gây lỗi? Giải thích mối liên hệ giữa định thức bằng 0 và việc ma trận không khả nghịch (singular matrix).

**A4 (Vận dụng cao)**

```python
import numpy as np
A = np.array([[2, 1], [1, 3]])
b = np.array([8, 13])
x = np.linalg.inv(A) @ b
x2 = np.linalg.solve(A, b)
print(x, x2)
```

Hỏi: Đoạn code trên đang giải bài toán gì (dạng `Ax = b`)? Hai cách tính `x` và `x2` có luôn cho kết quả giống nhau về mặt số học không? Cách nào được khuyến nghị dùng trong thực tế và tại sao (gợi ý: hiệu năng, độ ổn định số học khi ma trận lớn hoặc gần suy biến)?

## CHỦ ĐỀ 6: EIGENVALUES & EIGENVECTORS

### A. Đọc code – Trả lời tự luận

**A1 (Nhận biết)**

```python
import numpy as np
A = np.array([[4, 2], [1, 3]])
eigvals, eigvecs = np.linalg.eig(A)
print(eigvals)
print(eigvecs)
```

Hỏi: Hàm `np.linalg.eig` trả về mấy giá trị? Mỗi giá trị đó có ý nghĩa gì (trị riêng và vector riêng)?

**A2 (Thông hiểu)**

```python
import numpy as np
A = np.array([[4, 2], [1, 3]])
eigvals, eigvecs = np.linalg.eig(A)
v1 = eigvecs[:, 0]
lam1 = eigvals[0]
print(A @ v1)
print(lam1 * v1)
```

Hỏi: Hai kết quả in ra cuối cùng có quan hệ gì với nhau? Viết lại đẳng thức toán học định nghĩa trị riêng và vector riêng dựa trên quan sát này.

**A3 (Vận dụng)**

```python
import numpy as np
A = np.array([[2, 0], [0, 5]])
B = np.array([[1, 1], [0, 1]])
eigvals_A, _ = np.linalg.eig(A)
eigvals_B, _ = np.linalg.eig(B)
print(eigvals_A)
print(eigvals_B)
print(np.linalg.det(A), np.sum(eigvals_A))
print(np.linalg.det(B), np.sum(eigvals_B))
```

Hỏi: Với ma trận đường chéo `A`, trị riêng có liên hệ gì với các phần tử trên đường chéo? Từ dữ liệu in ra, hãy nêu 2 tính chất tổng quát liên hệ giữa trị riêng với định thức (det) và với vết ma trận (trace).

**A4 (Vận dụng cao)**

```python
import numpy as np
A = np.array([[0, -1], [1, 0]])
eigvals, eigvecs = np.linalg.eig(A)
print(eigvals)
```

## GỢI Ý ĐÁP ÁN NHANH (phần đọc code)

- **1.A1**: `C` = nhân từng phần tử (Hadamard product); `D` = nhân ma trận thật (row × column). Toán tử `*` là element-wise, `@` là matrix multiplication.
- **1.A2**: Điều kiện: số cột của A (3) = số hàng của B (3) → nhân được. Shape kết quả = (số hàng A, số cột B) = (2, 2).
- **1.A3**: `B = A.T` là chuyển vị của A. `C = A @ A.T` luôn đối xứng vì `(A A^T)^T = A^{T^T} A^T = A A^T`.
- **1.A4**: Khó tính nhẩm với B vì B không chéo hóa đơn giản qua lũy thừa từng phần tử. Ma trận đường chéo dễ lũy thừa vì `D^n` chỉ cần lũy thừa từng phần tử trên đường chéo, các phần tử ngoài đường chéo luôn là 0.
- **2.A1**: `inner` trả về 1 số vô hướng (scalar); `outer` trả về ma trận shape (3,3).
- **2.A2**: `c` vuông góc với cả `a` và `b` (theo quy tắc bàn tay phải). `cross` "tự nhiên" cho 3D vì tích có hướng chỉ định nghĩa chuẩn trong không gian 3 chiều.
- **2.A3**: Với input nhiều chiều, `np.inner` tính tổng theo trục cuối cùng của mỗi mảng, kết quả shape (2,2) — không phải phép nhân ma trận thông thường.
- **2.A4**: Mô phỏng mô-men lực và công cơ học; `torque = d × F`, `work = F · displacement`.
- **3.A1**: Công thức 2x2: `det = ad - bc`.
- **3.A2**: In ra ma trận đơn vị (gần đúng do làm tròn) vì `A @ A⁻¹ = I`.
- **3.A3**: `det(B) = 0` → B suy biến (các hàng/cột phụ thuộc tuyến tính) → không có nghịch đảo.
- **3.A4**: Giải `Ax = b`; hai cách cho kết quả gần giống nhau nhưng `solve` được khuyến nghị vì nhanh hơn và ổn định số học hơn (không cần tính nghịch đảo tường minh).
- **4.A1**: Kết quả = 1×0 + 2×1 + 3×0 = 2.
- **4.A2**: `outer` không yêu cầu cùng độ dài vì nó tạo ma trận từ mọi cặp phần tử; shape = (len(a), len(b)) = (2,3).
- **4.A3**: `trace(outer(x,x)) = inner(x,x) = sum(x_i^2)` — bằng nhau vì đường chéo của outer(x,x) chính là các `x_i * x_i`.
- **4.A4**: `M` có rank 1 (vì là outer product của 1 vector với chính nó) → chỉ có 1 trị riêng khác 0, các trị riêng còn lại xấp xỉ 0.
- **5.A1**: Giống nhau với số thực vì không có liên hợp phức để tạo khác biệt.
- **5.A2**: `vdot` lấy liên hợp phức của phần tử đầu tiên trước khi nhân, dùng trong định nghĩa tích vô hướng Hermitian; `dot` không lấy liên hợp.
- **5.A3**: `dot` tổng quát cho ma trận theo quy tắc nhân ma trận chuẩn; `vdot` luôn làm phẳng (flatten) input thành vector 1D trước khi tính tổng tích liên hợp — với 2 ma trận không cùng tổng số phần tử thì lỗi.
- **5.A4**: `vdot` tự động flatten `b` và `c` thành vector 1D có cùng 4 phần tử như `a`, nên kết quả trùng với `dot(a, a)`.
- **6.A1**: Trả về 2 giá trị: mảng trị riêng (eigenvalues) và ma trận vector riêng (mỗi cột là 1 eigenvector).
- **6.A2**: `A @ v1 == lam1 * v1`, đây chính là định nghĩa: `A v = λ v`.
- **6.A3**: Với ma trận chéo, trị riêng = các phần tử trên đường chéo. Tính chất tổng quát: `det(A) = tích các trị riêng`; `trace(A) = tổng các trị riêng`.
- **6.A4**: Phép xoay không có hướng nào bất biến (ngoại trừ góc đặc biệt) nên không tồn tại vector riêng thực với trị riêng thực — về mặt đại số, nghiệm của phương trình đặc trưng là số phức.
