```python
#Maximum Syntax
#numpy.maximum(arr1,arr2,/,out=None,*,...)
# keest quar trả về là 1 array với kích thước giống kích thước của 2 array ban đầu
# nếu 2 array ban đầu kích thước không giống nhau thì không thể so sánh được bằng maximum
# do đó ta có thể thêm np.nan vào phần tử cuối cùng, tuy nhiên khi 1 phần tử là nan thì kết quả maximum là nan

import numpy as np
a = [1,2,3,np.nan]
b = [2,4,1,2]
res = np.maximum(a,b)
print(res)

# tuy nhiên có thể so sánh 2x1 với 1x1 bằng kích cỡ
a=[[1,2,4],[5,4,3]]
b=[3,5,2]

res1 = np.maximum(a,b)
print(res1)

# min tương tự như max, syntax của min là: minimum 
```
