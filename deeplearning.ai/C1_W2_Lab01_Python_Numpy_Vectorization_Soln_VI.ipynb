### 3.4.1 Đánh chỉ mục (Indexing)
Các phần tử của vector có thể được truy cập thông qua đánh chỉ mục và cắt lát (slicing). NumPy cung cấp một bộ khả năng đánh chỉ mục và cắt lát rất đầy đủ. Chúng ta sẽ chỉ khám phá những kiến thức cơ bản cần thiết cho khóa học ở đây. Tham khảo [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html) để biết thêm chi tiết.  
**Đánh chỉ mục (Indexing)** nghĩa là tham chiếu đến *một phần tử* của mảng theo vị trí của nó trong mảng.  
**Cắt lát (Slicing)** nghĩa là lấy một *tập con* các phần tử từ một mảng dựa trên chỉ mục của chúng.  
NumPy bắt đầu đánh chỉ mục từ 0 nên phần tử thứ 3 của vector $\mathbf{a}$ là `a[2]`.

```python
#các phép toán đánh chỉ mục vector trên vector 1-D
a = np.arange(10)
print(a)

#truy cập một phần tử
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Truy cập một phần tử trả về một giá trị vô hướng (scalar)")

# truy cập phần tử cuối cùng, chỉ mục âm được đếm từ cuối
print(f"a[-1] = {a[-1]}")

#các chỉ mục phải nằm trong phạm vi của vector, nếu không chúng sẽ gây ra lỗi
try:
    c = a[10]
except Exception as e:
    print("Thông báo lỗi bạn sẽ thấy là:")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 Cắt lát (Slicing)
Cắt lát tạo ra một mảng các chỉ mục sử dụng một bộ ba giá trị (`start:stop:step`). Một tập con các giá trị cũng hợp lệ. Cách sử dụng của nó được giải thích tốt nhất qua ví dụ:

```python
#các phép toán cắt lát vector
a = np.arange(10)
print(f"a         = {a}")

#truy cập 5 phần tử liên tiếp (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# truy cập 3 phần tử cách nhau 2 vị trí
c = a[2:7:2];     print("a[2:7:2] = ", c)

# truy cập tất cả các phần tử từ chỉ mục 3 trở lên
c = a[3:];        print("a[3:]    = ", c)

# truy cập tất cả các phần tử dưới chỉ mục 3
c = a[:3];        print("a[:3]    = ", c)

# truy cập tất cả các phần tử
c = a[:];         print("a[:]     = ", c)
```

### 3.4.6 Tích vô hướng (Dot product) giữa Vector và Vector
Tích vô hướng là một trụ cột của Đại số Tuyến tính và NumPy. Đây là một phép toán được sử dụng rộng rãi trong khóa học này và cần được hiểu rõ. Tích vô hướng được thể hiện bên dưới.

<img src="./images/C1_W2_Lab04_dot_notrans.gif" width=800> 

Tích vô hướng nhân các giá trị trong hai vector theo từng phần tử rồi cộng tổng kết quả lại.
Tích vô hướng của vector yêu cầu chiều của hai vector phải giống nhau.

Hãy triển khai phiên bản tích vô hướng của riêng chúng ta bên dưới:

**Sử dụng vòng lặp for**, triển khai một hàm trả về tích vô hướng của hai vector. Hàm trả về với đầu vào $a$ và $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Giả sử cả `a` và `b` đều có cùng kích thước.

```python
def my_dot(a, b): 
    """
   Tính tích vô hướng của hai vector
 
    Args:
      a (ndarray (n,)):  vector đầu vào 
      b (ndarray (n,)):  vector đầu vào có cùng chiều với a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# kiểm thử 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Lưu ý, tích vô hướng được kỳ vọng trả về một giá trị vô hướng. 

Hãy thử cùng phép toán này bằng cách sử dụng `np.dot`.

```python
# kiểm thử 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

Ở trên, bạn sẽ nhận thấy rằng kết quả cho 1-D khớp với triển khai của chúng ta.

### 3.4.7 Cần tốc độ: vector hóa so với vòng lặp for
Chúng ta sử dụng thư viện NumPy vì nó cải thiện hiệu suất bộ nhớ và tốc độ. Hãy chứng minh điều đó:

```python
np.random.seed(1)
a = np.random.rand(10000000)  # các mảng rất lớn
b = np.random.rand(10000000)

tic = time.time()  # ghi lại thời gian bắt đầu
c = np.dot(a, b)
toc = time.time()  # ghi lại thời gian kết thúc

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Thời gian phiên bản vector hóa: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # ghi lại thời gian bắt đầu
c = my_dot(a,b)
toc = time.time()  # ghi lại thời gian kết thúc

print(f"my_dot(a, b) =  {c:.4f}")
print(f"Thời gian phiên bản vòng lặp: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #xóa các mảng lớn này khỏi bộ nhớ
```

Vậy, vector hóa mang lại sự tăng tốc lớn trong ví dụ này. Điều này là vì NumPy tận dụng tốt hơn khả năng song song hóa dữ liệu (data parallelism) có sẵn trong phần cứng bên dưới. Cấu trúc dữ liệu cơ bản của NumPy là một *mảng* n-chiều có thể đánh chỉ mục, chứa các phần tử cùng kiểu (`dtype`). Điều này đã được mô tả trước đó. Ma trận có chỉ mục hai chiều (2-D) [m,n].

Trong Khóa học 1, ma trận 2-D được sử dụng để lưu trữ dữ liệu huấn luyện. Dữ liệu huấn luyện có $m$ ví dụ với $n$ đặc trưng tạo thành một mảng (m,n). Khóa học 1 không thực hiện các phép toán trực tiếp trên ma trận mà thường trích xuất một ví dụ dưới dạng vector và thực hiện phép toán trên đó. Bên dưới bạn sẽ ôn lại: 
- tạo dữ liệu
- cắt lát và đánh chỉ mục

## 4.3 Tạo Ma trận
Các hàm tương tự tạo ra vector 1-D sẽ tạo ra mảng 2-D hoặc n-D. Dưới đây là một số ví dụ

Bên dưới, tuple kích thước được cung cấp để đạt được kết quả 2-D. Chú ý cách NumPy sử dụng dấu ngoặc vuông để biểu thị mỗi chiều. Chú ý thêm rằng NumPy, khi in, sẽ in mỗi hàng trên một dòng.

```python
a = np.zeros((1, 5))                                       
print(f"a shape = {a.shape}, a = {a}")                     

a = np.zeros((2, 1))                                                                   
print(f"a shape = {a.shape}, a = {a}") 

a = np.random.random_sample((1, 1))  
print(f"a shape = {a.shape}, a = {a}") 
```

Người ta cũng có thể chỉ định dữ liệu thủ công. Các chiều được chỉ định bằng dấu ngoặc bổ sung khớp với định dạng in ở trên.

```python
# Các hàm NumPy cấp phát bộ nhớ và điền các giá trị do người dùng chỉ định
a = np.array([[5], [4], [3]]);   print(f" a shape = {a.shape}, np.array: a = {a}")
a = np.array([[5],   # Người ta cũng có thể
              [4],   # tách các giá trị
              [3]]); #thành các hàng riêng biệt
print(f" a shape = {a.shape}, np.array: a = {a}")
```

## 4.4 Các phép toán trên Ma trận
Hãy khám phá một số phép toán sử dụng ma trận.

<a name="toc_40015_4.4.1"></a>
### 4.4.1 Đánh chỉ mục (Indexing)

Ma trận có thêm một chỉ mục thứ hai. Hai chỉ mục mô tả [hàng, cột]. Việc truy cập có thể trả về một phần tử hoặc một hàng/cột. Xem bên dưới:

```python
#các phép toán đánh chỉ mục vector trên ma trận
a = np.arange(6).reshape(-1, 2)   #reshape là một cách thuận tiện để tạo ma trận
print(f"a.shape: {a.shape}, \na= {a}")

#truy cập một phần tử
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Truy cập một phần tử trả về một giá trị vô hướng\n")

#truy cập một hàng
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

Đáng chú ý là ví dụ cuối cùng. Việc truy cập một ma trận chỉ bằng cách chỉ định hàng sẽ trả về một *vector 1-D*.

**Reshape**  
Ví dụ trước đã sử dụng [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) để định hình mảng.  
`a = np.arange(6).reshape(-1, 2) `   
Dòng mã này trước tiên tạo ra một *Vector 1-D* gồm sáu phần tử. Sau đó nó định hình lại vector đó thành một mảng *2-D* bằng lệnh reshape. Điều này có thể được viết là:  
`a = np.arange(6).reshape(3, 2) `  
Để đạt được cùng một mảng 3 hàng, 2 cột.
Tham số -1 báo cho hàm tự tính toán số hàng dựa trên kích thước của mảng và số cột.

<a name="toc_40015_4.4.2"></a>
### 4.4.2 Cắt lát (Slicing)
Cắt lát tạo ra một mảng các chỉ mục sử dụng một bộ ba giá trị (`start:stop:step`). Một tập con các giá trị cũng hợp lệ. Cách sử dụng của nó được giải thích tốt nhất qua ví dụ:

```python
#các phép toán cắt lát vector 2-D
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#truy cập 5 phần tử liên tiếp (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "một mảng 1-D")

#truy cập 5 phần tử liên tiếp (start:stop:step) trong hai hàng
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "một mảng 2-D")

# truy cập tất cả các phần tử
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# truy cập tất cả các phần tử trong một hàng (cách dùng rất phổ biến)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "một mảng 1-D")
# giống như
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "một mảng 1-D")
```
