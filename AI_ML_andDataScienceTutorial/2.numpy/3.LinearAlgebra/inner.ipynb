```python
# tính định thức sử dụng numpy.linalg.det()
# tức là khi sử dụng các thứ liên quan tới ma trận như định thức, nghịch đảo thì luốn ử dụng numpy.linalg.
# sau đó định thức thì numpy.linalg.det()
# nghịch đảo là: numpy.linalg.inv()

import numpy as np
A = np.array([[1,2],[3,4]])
res= np.linalg.det(A)
print(res)

# xác định dấu sử dụng linalg.slogdet
sign,logdet = np.linalg.slogdet(A)
print(sign, logdet)

# tính ma trận nghịch đảo
A_1 = np.linalg.inv(A)
res2 = np.dot(A,A_1)
print(res2)
```
